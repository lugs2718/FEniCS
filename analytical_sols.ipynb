# Solução analítica simbólica para o problema homogeneizado

```python
import sympy as sp

# Definições
r = sp.symbols('r', positive=True)
u = sp.Function('u')(r)

# Equação diferencial (sem o fator C_11, que é só constante)
eq = sp.Eq(sp.diff(u, r, 2) + (1/r)*sp.diff(u, r) - u/r**2, 0)

# Resolver simbolicamente
sol = sp.dsolve(eq)
print("Solução geral:")
display(sol)

# Exemplo: aplicando condições de contorno
# Digamos que u(a) = u_a e u(b) = u_b
a, b, u_a, u_b = sp.symbols('a b u_a u_b')
C1, C2 = sp.symbols('C1 C2')

# A solução geral é da forma u(r) = C1*r + C2/r
u_general = C1*r + C2/r

# Sistema com condições de contorno
eqs = [
    sp.Eq(u_general.subs(r, a), u_a),
    sp.Eq(u_general.subs(r, b), u_b)
]

# Resolver para C1 e C2
consts = sp.solve(eqs, [C1, C2])
u_solution = u_general.subs(consts)
u_solution = sp.simplify(u_solution)

print("Solução particular com Dirichlet:")
display(sp.simplify(u_solution))

# u_solution = u_solution.subs(u_a, 1)
# u_solution = sp.lambdify(r, u_solution, 'numpy')

```
